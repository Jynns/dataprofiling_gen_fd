```python
def find_fds_for_lhs(target_lhs, data, validator_rhs = lambda x : True ):
    """filters based on rhs and allows to sort and ensure min and max on lhs"""
    def shouldInclude(lhs_rhs):
        lhs, rhs = lhs_rhs
        return set(lhs) == set(target_lhs) and validator_rhs(rhs)
    return list(filter(shouldInclude, data))
```

```python
genuine_fds = find_fds_for_lhs(['CustomerID'], data, validator_rhs = lambda x: x.startswith('cust_') )
```

```python
ungenuine_fds = find_fds_for_lhs(['CustomerID'], data, validator_rhs = lambda x: x.startswith('ord_') )
```

```python
def fuzzy_filter_rhs(data, f):
    def unpack(lhs_rhs):
        lhs, rhs = lhs_rhs
        return f(rhs)
    return list(filter(unpack,data))
```

```python
fuzzy_filter_rhs(data, lambda rhs: rhs.startswith('cust_'))
ungenuine_fds.extend(fuzzy_filter(fuzzy_filter_rhs(data, lambda rhs: rhs.startswith('cust_')))[:5])
```
